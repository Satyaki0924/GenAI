# Tokenization with NLP
Tokenization is the fundamental first step in Natural Language Processing (NLP). It's the process of converting a large piece of text into smaller, meaningful units that a machine can easily process.

## Declare a corpus
A corpus is a paragraph of words

```python
corpus = """Satyaki's notes: Natural language processing (NLP) is a subfield of artificial intelligence (AI) that focuses on the interaction between computers and humans through natural language. The ultimate objective of NLP is to enable computers to understand, interpret, and generate human language in a way that is valuable. NLP combines computational linguistics—rule-based modeling of human language—with statistical, machine learning, and deep learning models. These technologies enable computers to process human language in the form of text or voice data and to 'understand' its full meaning, complete with the speaker or writer's intent and sentiment."""
```

## A. Document Tokenization
A document is a single piece of text within the corpus like a sentence.

```python
from nltk.tokenize import sent_tokenize
documents = sent_tokenize(corpus)
print("Document Tokenization: ", documents)
print ("Length of documents:", len(documents))
print("Type of documents:", type(documents))
```

## B. Word Tokenization
A Word is the basic unit of a language. A group of word creates sentences which in turn make a corpus

```python
from nltk.tokenize import word_tokenize
words = word_tokenize(corpus)
print("Word Tokenization: ", words)
print ("Length of words:", len(words))
print("Type of words:", type(words))
```

```python
# Word Tokenization of documents
for doc in documents:
    print(f"Word Tokenization of document: {doc}")
    words_in_doc = word_tokenize(doc)
    print("Words: ", words_in_doc)
    print("Length of words in document:", len(words_in_doc)) 
```

### Other Tokenizers
Apart from sent_tokenize and word_tokenize, there are other types of tokenizers too.

WordsPunktTokenizer, TreebankWordTokenizer, TweetTokenizer, RegexpTokenizer etc. are some of the other tokenizers available in nltk.

```python
"""
word_tokenize is a wrapper around NLTK’s TreebankWordTokenizer 
(linguistic rules). It treats contractions and many clitics as 
single logical tokens or as a token+clitic pair 
(e.g. "don't" → ["Do", "n't"]). 
It is more linguistically informed and is a good default for NLP tasks.
WordPunctTokenizer simply splits on punctuation (using a regex). 
It separates punctuation into its own tokens 
(e.g. "don't" → ["don", "'", "t"]). It’s simpler and more “mechanical”.
"""
from nltk.tokenize import WordPunctTokenizer
word_punkt = WordPunctTokenizer()
words_punkt_tokens = word_punkt.tokenize(corpus)
print("WordsPunktTokenizer Tokens: ", words_punkt_tokens)
print("Length of WordsPunktTokenizer tokens:", len(words_punkt_tokens))
```

```python
## TreeBankWordTokenizer is the same as word_tokenize
from nltk.tokenize import word_tokenize, WordPunctTokenizer

s = "Don't stop believing—hold on to that feelin'! State-of-the-art tech."

print("Treebank / word_tokenize:")
print(word_tokenize(s))

print("\nWordPunctTokenizer:")
print(WordPunctTokenizer().tokenize(s))
```
